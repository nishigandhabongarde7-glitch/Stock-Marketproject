```python
df = df.drop_duplicates()
```

```python
df['Year'] = df['Date'].dt.year
```

```python
df['Month'] = df['Date'].dt.month
```

```python
df['Month_Name'] = df['Date'].dt.month_name()
```

```python
df[['Date', 'Year', 'Month', 'Month_Name']].head()
```

```python
df = df.sort_values(['Company', 'Date'])
```

```python
df['Daily_Return'] = df.groupby('Company')['Close'].pct_change() * 100
```

```python
df[['Company', 'Date', 'Close', 'Daily_Return']].head(10)
```

```python
df['Daily_Return'].describe()
```

```python
df['Price_Range'] = df['High'] - df['Low']
```

```python
df[['Company', 'Date', 'High', 'Low', 'Price_Range']].head(10)
```

```python
import numpy as np
```

```python
df['Price_Movement'] = np.where(
    df['Price_Range'] > df['Price_Range'].median(),
    'High Movement',
    'Low Movement'
)
```

```python
df[['Company', 'Date', 'Price_Range', 'Price_Movement']].head(10)
```

```python
df['Price_Movement'].value_counts()
```

```python
company_avg_close = df.groupby('Company')['Close'].mean().sort_values(ascending=False)
```

```python
company_avg_close = company_avg_close.reset_index()
```

```python
company_avg_close.columns = ['Company', 'Average_Close']
```

```python
company_avg_close['Average_Close'] = np.round(
    company_avg_close['Average_Close'], 2
)
```

```python
company_avg_close
```

```python
company_avg_return = df.groupby('Company')['Daily_Return'].mean().sort_values(ascending=False)
```

```python
company_avg_return = company_avg_return.reset_index()
company_avg_return.columns = ['Company', 'Average_Daily_Return']
```

```python
company_avg_return['Average_Daily_Return'] = np.round(
    company_avg_return['Average_Daily_Return'], 2
)
```

```python
company_avg_return
```

```python
company_summary = pd.merge(
    company_avg_close,
    company_avg_return,
    on='Company'
)
```

```python
company_summary
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    company_summary['Company'],
    company_summary['Average_Close']
)

plt.xlabel('Company')
plt.ylabel('Average Closing Price')
plt.title('Company-wise Average Closing Price')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
